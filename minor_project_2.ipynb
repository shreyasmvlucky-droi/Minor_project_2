```python
def extract_vendor(description):

    description = str(description).upper()

    # ATM withdrawals
    if description.startswith("ATM-WDL"):
        return "Cash Withdrawal"

    # Friend-to-friend transfers
    p2p_names = [
        "AMAN",
        "ANKIT",
        "PRIYA",
        "VIKAS",
        "NEHA",
        "SNEHA",
        "KARAN"
    ]

    if description.startswith("UPI-"):
        for name in p2p_names:
            if name in description:
                return "P2P Transfer"

    # Rent / landlord transfers
    if "RENT-LANDLORD" in description:
        return "Personal Transfer"

    # Vendor matching
    for vendor, keywords in vendor_keywords.items():

        for keyword in keywords:

            if keyword in description:
                return vendor

    # If nothing matches
    return "Uncategorised"
```

```python
df["vendor_clean"] = df["Description"].apply(extract_vendor)
```

```python
growth = {}

for category in monthly_spending.index:

    first_month = monthly_spending.loc[category, "Jan"]
    last_month = monthly_spending.loc[category, "Jun"]

    if first_month != 0:
        growth[category] = (
            (last_month - first_month) / first_month
        ) * 100

growth = pd.Series(growth).sort_values(ascending=False)

print("Jan to Jun growth:")
print(growth)
```

## Feature 6 — Time-of-Day Patterns.

```python
df["hour"] = df["Time"].str[:2].astype(int)

print(df[["Time", "hour"]].head(10))
```

```python
def get_time_period(hour):

    if 5 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour < 21:
        return "Evening"

    else:
        return "Night"
```

```python
df["time_period"] = df["hour"].apply(get_time_period)
```

```python
print(df["time_period"].value_counts())
```

```python
time_matrix = df[df["type_clean"] == "debit"].pivot_table(
    values="amount",
    index="category",
    columns="hour",
    aggfunc="sum",
    fill_value=0
)
```

```python
time_matrix = time_matrix.reindex(
    columns=range(24),
    fill_value=0
)
```

```python
print(time_matrix)
```

```python
hourly_spending = time_matrix.sum(axis=0)

busiest_hour = hourly_spending.idxmax()
busiest_hour_amount = hourly_spending.max()

print("Busiest spending hour:",
      busiest_hour)

print("Amount spent:",
      busiest_hour_amount)
```

```python
# Feature 8 - Spending Archetype Detection

total_debits = df.loc[
    df["type_clean"] == "debit",
    "amount"
].sum()

print("Total debits:", total_debits)
```

```python
# Late-Night Snacker

food_delivery = df[
    (df["type_clean"] == "debit") &
    (df["category"] == "Food Delivery")
]

total_food_delivery_transactions = len(food_delivery)

late_night_food = food_delivery[
    (food_delivery["hour"] >= 21) |
    (food_delivery["hour"] <= 2)
]

late_night_percentage = (
    len(late_night_food) /
    total_food_delivery_transactions
) * 100

print("Total Food Delivery transactions:",
      total_food_delivery_transactions)

print("Late-night Food Delivery transactions:",
      len(late_night_food))

print(f"Late-night percentage: {late_night_percentage:.2f}%")
```

```python
subscription_vendors = df[
    (df["type_clean"] == "debit") &
    (df["category"] == "Subscriptions")
]["vendor_clean"].dropna().unique()

print("Subscription vendors:")
print(subscription_vendors)

print("\nNumber of distinct subscription vendors:",
      len(subscription_vendors))
```

```python
print("=" * 75)
print("FEATURE 8 - SPENDING ARCHETYPE DETECTION")
print("=" * 75)

archetypes = {
    "THE FOODIE": (
        category_percentage.get("Food Delivery", 0)
        + category_percentage.get("Restaurants", 0)
        + category_percentage.get("Cafe", 0)
    ),

    "THE QUICK COMMERCE JUNKIE":
        category_percentage.get("Quick Commerce", 0),

    "THE SHOPAHOLIC":
        category_percentage.get("E-commerce", 0),

    "THE INVESTOR":
        category_percentage.get("Investments", 0),

    "THE LATE-NIGHT SNACKER":
        late_night_percentage,

    "THE CAB COMMUTER":
        category_percentage.get("Transport", 0),

    "THE SUBSCRIPTION LOVER":
        len(subscription_vendors),

    "THE YOLO SPENDER":
        savings_rate,

    "THE DISCIPLINED SAVER":
        savings_rate
}

# Apply the rules
results = {
    "THE FOODIE":
        archetypes["THE FOODIE"] > 25,

    "THE QUICK COMMERCE JUNKIE":
        archetypes["THE QUICK COMMERCE JUNKIE"] > 15,

    "THE SHOPAHOLIC":
        archetypes["THE SHOPAHOLIC"] > 15,

    "THE INVESTOR":
        archetypes["THE INVESTOR"] > 15,

    "THE LATE-NIGHT SNACKER":
        archetypes["THE LATE-NIGHT SNACKER"] > 50,

    "THE CAB COMMUTER":
        archetypes["THE CAB COMMUTER"] > 10,

    "THE SUBSCRIPTION LOVER":
        archetypes["THE SUBSCRIPTION LOVER"] >= 5,

    "THE YOLO SPENDER":
        archetypes["THE YOLO SPENDER"] < 10,

    "THE DISCIPLINED SAVER":
        archetypes["THE DISCIPLINED SAVER"] > 40
}

print("\nArchetype Results:")

for name, applies in results.items():

    metric = archetypes[name]

    if name == "THE SUBSCRIPTION LOVER":
        print(
            f"{name}: "
            f"{'YES' if applies else 'NO'} "
            f"({metric} distinct vendors)"
        )
    else:
        print(
            f"{name}: "
            f"{'YES' if applies else 'NO'} "
            f"({metric:.2f}%)"
        )

print("\n" + "=" * 75)
print("ARCHETYPES THAT APPLY TO RAHUL")
print("=" * 75)

matched = [
    name for name, applies in results.items()
    if applies
]

for name in matched:
    metric = archetypes[name]

    if name == "THE SUBSCRIPTION LOVER":
        print(f"{name} → {metric} distinct vendors")
    else:
        print(f"{name} → {metric:.2f}%")

print("=" * 75)
```

```python
# ============================================================
#                 FINAL SPENDDNA REPORT
# ============================================================

print("=" * 70)
print("                 SpendDNA REPORT")
print("=" * 70)

# ------------------------------------------------------------
# 1. EXECUTIVE SUMMARY
# ------------------------------------------------------------

total_credits = df.loc[
    df["type_clean"] == "credit", "amount"
].sum()

total_debits = df.loc[
    df["type_clean"] == "debit", "amount"
].sum()

net_savings = total_credits - total_debits

savings_rate = (
    net_savings / total_credits * 100
    if total_credits != 0 else 0
)

transaction_count = len(df)

unique_vendors = df["vendor_clean"].nunique()

print("\nEXECUTIVE SUMMARY")
print("-" * 70)

print(f"Total credits       : Rs. {total_credits:,.2f}")
print(f"Total debits        : Rs. {total_debits:,.2f}")
print(f"Net savings         : Rs. {net_savings:,.2f}")
print(f"Savings rate        : {savings_rate:.2f}%")
print(f"Transactions         : {transaction_count}")
print(f"Unique vendors      : {unique_vendors}")


# ------------------------------------------------------------
# 2. TOP 5 CATEGORIES
# ------------------------------------------------------------

debit_df = df[df["type_clean"] == "debit"].copy()

category_spending = (
    debit_df
    .groupby("category")["amount"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTOP 5 CATEGORIES")
print("-" * 70)

for category, amount in category_spending.head(5).items():

    percentage = amount / total_debits * 100

    print(
        f"{category:<20} "
        f"{percentage:>6.2f}%   "
        f"Rs. {amount:>10,.2f}"
    )


# ------------------------------------------------------------
# 3. TOP 5 VENDORS
# ------------------------------------------------------------

vendor_spending = (
    debit_df
    .groupby("vendor_clean")["amount"]
    .agg(["sum", "count"])
    .sort_values("sum", ascending=False)
)

print("\nTOP 5 VENDORS")
print("-" * 70)

for vendor, row in vendor_spending.head(5).iterrows():

    print(
        f"{vendor:<20} "
        f"Rs. {row['sum']:>10,.2f} "
        f"({int(row['count'])} orders)"
    )


# ------------------------------------------------------------
# 4. TIME-OF-DAY PATTERNS
# ------------------------------------------------------------

# Extract hour from Time column
df["hour"] = (
    df["Time"]
    .astype(str)
    .str[:2]
    .astype(int)
)

food_delivery = debit_df[
    debit_df["category"] == "Food Delivery"
]

food_delivery_hours = (
    food_delivery["Time"]
    .astype(str)
    .str[:2]
    .astype(int)
)

late_night_food = food_delivery[
    (food_delivery_hours >= 21) |
    (food_delivery_hours <= 2)
]

late_night_percentage = (
    len(late_night_food) / len(food_delivery) * 100
    if len(food_delivery) > 0 else 0
)

# Cafe peak hour
cafe_df = debit_df[
    debit_df["category"] == "Cafe"
]

if len(cafe_df) > 0:

    cafe_hours = (
        cafe_df["Time"]
        .astype(str)
        .str[:2]
        .astype(int)
    )

    cafe_peak_hour = cafe_hours.value_counts().idxmax()

else:
    cafe_peak_hour = None


# Quick Commerce dominant time period
quick_df = debit_df[
    debit_df["category"] == "Quick Commerce"
].copy()


def get_time_period(hour):

    if 5 <= hour < 12:
        return "Morning"

    elif 12 <= hour < 17:
        return "Afternoon"

    elif 17 <= hour < 21:
        return "Evening"

    else:
        return "Night"


if len(quick_df) > 0:

    quick_hours = (
        quick_df["Time"]
        .astype(str)
        .str[:2]
        .astype(int)
    )

    quick_periods = quick_hours.apply(get_time_period)

    quick_peak_period = quick_periods.value_counts().idxmax()

else:
    quick_peak_period = "No data"


print("\nTIME-OF-DAY PATTERNS")
print("-" * 70)

print(
    f"Food Delivery late-night : "
    f"{late_night_percentage:.2f}% "
    f"(21:00 - 02:00)"
)

if cafe_peak_hour is not None:

    print(
        f"Cafe peak hour          : "
        f"{cafe_peak_hour:02d}:00"
    )

print(
    f"Quick Commerce pattern   : "
    f"{quick_peak_period}"
)


# ------------------------------------------------------------
# 5. MONTHLY TREND - FOOD DELIVERY
# ------------------------------------------------------------

df["month_num"] = df["date"].dt.month

month_names = {
    1: "Jan",
    2: "Feb",
    3: "Mar",
    4: "Apr",
    5: "May",
    6: "Jun"
}

food_monthly = (
    df[
        (df["type_clean"] == "debit") &
        (df["category"] == "Food Delivery")
    ]
    .groupby("month_num")["amount"]
    .sum()
)

print("\nMONTHLY TREND - FOOD DELIVERY")
print("-" * 70)

for month_num in range(1, 7):

    amount = food_monthly.get(month_num, 0)

    print(
        f"{month_names[month_num]} "
        f"Rs. {amount:,.2f}"
    )

# ------------------------------------------------------------
# 6. ANOMALY DETECTION
# ------------------------------------------------------------

# Category mean and standard deviation
category_mean = (
    debit_df
    .groupby("category")["amount"]
    .transform("mean")
)

category_std = (
    debit_df
    .groupby("category")["amount"]
    .transform("std")
)

debit_df["z_score"] = (
    (debit_df["amount"] - category_mean)
    / category_std
)

anomalies = debit_df[
    debit_df["z_score"] > 2
].sort_values(
    "z_score",
    ascending=False
)

print("\nTOP 5 ANOMALIES")
print("-" * 70)

for _, row in anomalies.head(5).iterrows():

    date_value = row["date"]

    if pd.notna(date_value):
        date_text = date_value.strftime("%d %b")
    else:
        date_text = "Unknown"

    print(
        f"{date_text:<10} "
        f"{row['vendor_clean']:<18} "
        f"Rs. {row['amount']:>9,.2f} "
        f"(z={row['z_score']:.2f})"
    )

print(f"\nTotal anomalies: {len(anomalies)}")


# ------------------------------------------------------------
# 7. SPENDING ARCHETYPES
# ------------------------------------------------------------

category_percentage = (
    category_spending / total_debits * 100
)

foodie_percentage = (
    category_percentage.get("Food Delivery", 0)
    + category_percentage.get("Restaurants", 0)
    + category_percentage.get("Cafe", 0)
)

quick_percentage = (
    category_percentage.get("Quick Commerce", 0)
)

ecommerce_percentage = (
    category_percentage.get("E-commerce", 0)
)

investment_percentage = (
    category_percentage.get("Investments", 0)
)

transport_percentage = (
    category_percentage.get("Transport", 0)
)

subscription_vendors = debit_df[
    debit_df["category"] == "Subscriptions"
]["vendor_clean"].dropna().unique()

subscription_count = len(subscription_vendors)


archetypes = []

if foodie_percentage > 25:
    archetypes.append(
        ("THE FOODIE", f"{foodie_percentage:.2f}% on food")
    )

if quick_percentage > 15:
    archetypes.append(
        (
            "THE QUICK COMMERCE JUNKIE",
            f"{quick_percentage:.2f}% on Quick Commerce"
        )
    )

if ecommerce_percentage > 15:
    archetypes.append(
        (
            "THE SHOPAHOLIC",
            f"{ecommerce_percentage:.2f}% on E-commerce"
        )
    )

if investment_percentage > 15:
    archetypes.append(
        (
            "THE INVESTOR",
            f"{investment_percentage:.2f}% on Investments"
        )
    )

if late_night_percentage > 50:
    archetypes.append(
        (
            "THE LATE-NIGHT SNACKER",
            f"{late_night_percentage:.2f}% of Food Delivery transactions late-night"
        )
    )

if transport_percentage > 10:
    archetypes.append(
        (
            "THE CAB COMMUTER",
            f"{transport_percentage:.2f}% on Transport"
        )
    )

if subscription_count >= 5:
    archetypes.append(
        (
            "THE SUBSCRIPTION LOVER",
            f"{subscription_count} subscription vendors"
        )
    )

if savings_rate < 10:
    archetypes.append(
        (
            "THE YOLO SPENDER",
            f"savings rate {savings_rate:.2f}%"
        )
    )

if savings_rate > 40:
    archetypes.append(
        (
            "THE DISCIPLINED SAVER",
            f"savings rate {savings_rate:.2f}%"
        )
    )


print("\nRAHUL'S SPENDING ARCHETYPES")
print("-" * 70)

for name, metric in archetypes:

    print(
        f"-> {name:<30} ({metric})"
    )


# ------------------------------------------------------------
# 8. KEY INSIGHTS
# ------------------------------------------------------------

top_category = category_spending.index[0]
top_category_amount = category_spending.iloc[0]

top_vendor = vendor_spending.index[0]
top_vendor_amount = vendor_spending.iloc[0]["sum"]

print("\nKEY INSIGHTS")
print("-" * 70)

print(
    f"1. Rahul's spending is highest in "
    f"{top_category}, with Rs. {top_category_amount:,.2f} spent."
)

print(
    f"2. {top_vendor} is the highest-spend vendor, "
    f"with Rs. {top_vendor_amount:,.2f} spent."
)

print(
    f"3. Rahul's savings rate is {savings_rate:.2f}%, "
    f"indicating an overspending pattern."
)

print(
    f"4. {late_night_percentage:.2f}% of Food Delivery "
    f"transactions occur between 21:00 and 02:00."
)

print(
    f"5. The analysis identified {len(anomalies)} "
    f"anomalous transactions."
)

print("\n" + "=" * 70)
print("                    END OF REPORT")
print("=" * 70)
```
